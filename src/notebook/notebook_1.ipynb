En este notebook se construye el flujo de trabajo completo para la primera parte del proyecto carga y creación de el dataset llamado "California housing" o en español "Viviendas de California".


### **1. Se prepara el entorno e importa la clase de carga**

```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
while ROOT != ROOT.parent and not (ROOT / "src" / "clases").exists():
    ROOT = ROOT.parent

sys.path.insert(0, str(ROOT / "src"))

from clases import HFDatasetLoader
from clases import Reportes
```

### **2. Se configura el dataset y la salida**

```python
loader = HFDatasetLoader( dataset_name="BrianSuToronto/california_housing", split="train", filename= "data/california_housing.csv")
```

### **3. Se descarga el dataset y se crea el DataFrame**

```python
df = loader.load()
df.head(3)
```

### **4. Revisamos la estructura del DataFrame**

```python
print(f"Filas: {len(df)}")
print(f"Columnas: {len(df.columns)}")
df.info()
```

### **5. Apartir de aquí hacemos las transformaciones mínimas**

#### **Tranformación 1 (Renombrado)**

```python
# Mostramos los nombres originales de las columnas
print(df.columns)
```

```python
# Renombramos las columnas
df_transformado = df.rename(columns={
    "MedInc": "Ingreso_mediano",
    "HouseAge": "Edad_vivienda",
    "AveRooms": "Promedio_habitaciones",
    "AveBedrms": "Promedio_dormitorio_hogar",
    "Population": "Poblacion",
    "AveOccup": "Promedio_ocupacion",
    "Latitude": "Latitud",
    "Longitude": "Longitud",
    "MedHouseVal": "Valor_mediano_vivienda"
})
```

#### **Transformación 2 (Filtrado)**

```python
df_t2 = df_transformado[df_transformado["Edad_vivienda"] > 20]
```

#### **Transformación 3 (Seleccionado)**

```python
df_t3 = df_t2[[
    "Ingreso_mediano",
    "Edad_vivienda",
    "Promedio_habitaciones",
    "Poblacion",
    "Valor_mediano_vivienda",
]]
```

#### **Transformación 4 (Columna nueva)**

```python
df_t4 = df_t3.assign(Valor_en_dolares=df_t3["Valor_mediano_vivienda"] * 100000)
```

### **6. Resultados y explicaciones de las transformaciones**

**Resultados por filas y columas**

```python
# Se muestran los resultados por filas y columnas
print("Se muestran los resultados por filas y columnas")
print("Original:     ", df.shape)
print("Renombrado:   ", df_transformado.shape)
print("Filtrado:     ", df_t2.shape)
print("Seleccionado: ", df_t3.shape)
print("Con columna:  ", df_t4.shape)
print(df_t4.columns.tolist())
```

**Resultados de cada transformación**

```python
# Importamos display para mostrar los resultados de cada transformación 
from IPython.display import display
# Nombres de las columnas originales
print("Original (nombres en inglés):")
display(df.head())
# Columnas renombradas
print("Renombrado (nombres en español):")
display(df_transformado.head())
# Filtrado 
print("Filtrado (Edad_vivienda > 20):")
display(df_t2.head())
# Seleccionado 
print("Seleccionado (5 columnas):")
display(df_t3.head())
# Nueva columna
print("Con la nueva columna Valor_en_dolares:")
display(df_t4.head())
```

#### **Explicación 1 (Renombrado)**
Se usó `DataFrame.rename()` para cambiar los nombres de las columnas de inglés a español (por ejemplo, `MedInc` pasó a `Ingreso_mediano`), porque los nombres en español son más fáciles de entender y tienen una mejor interpretación en todo el análisis, el resultado es el DataFrame conservando las mismas 20640 filas y 9 columnas, y solo cambiaron los nombres. El DataFrame original (`df`) no se modificó, porque `rename()` devuelve uno nuevo, guardado en `df_transformado`.

#### **Explicación 2 (Filtrado)**
Se filtró `df_transformado` con la condición `Edad_vivienda > 20`, y el resultado se guardó en `df_t2`así se analizan solo las zonas con viviendas de más de 20 años de antigüedad mediana y el DataFrame pasó de 20640 a 14347 filas y mantiene las 9 columnas. Las filas que no cumplían la condición se descartaron.

#### **Explicación 3 (Seleccionado)**
A partir de `df_t2` se seleccionaron cinco columnas (`Ingreso_mediano`, `Edad_vivienda`, `Promedio_habitaciones`, `Poblacion` y `Valor_mediano_vivienda`), y el resultado se guardó en `df_t3`, se conservan solo las variables más relevantes y se descartan las demás para simplificar el análisis y el número de filas se mantiene en 14347 y las columnas bajaron de 9 a 5.

#### **Explicación 4 (Columna nueva)**
Con `assign()` se creó la columna `Valor_en_dolares`, que multiplica `Valor_mediano_vivienda` por 100.000. El resultado se guardó en `df_t4`, porque en este dataset el valor mediano de la vivienda está expresado en cientos de miles de dólares, por lo que la nueva columna lo muestra en dólares y es más fácil de interpretar (por ejemplo, 3.5 equivale a 350.000 USD) y el DataFrame mantiene 14347 filas y pasó de 5 a 6 columnas.

#### **Resultado final y reporte**

El DataFrame final (`df_t4`) tiene 14347 filas y 6 columnas. Este DataFrame transformado se pasa a la clase `Reportes`, que generó el archivo `.txt` con el resumen: filas, columnas, duplicados, tipos de dato, valores nulos y estadísticas numéricas. Por eso, el archivo corresponde a los datos transformados y no a los originales.
