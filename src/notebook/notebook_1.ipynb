### **1. Prepara el entorno e importa la clase de carga**

```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
while ROOT != ROOT.parent and not (ROOT / "src" / "clases").exists():
    ROOT = ROOT.parent

sys.path.insert(0, str(ROOT / "src"))

from clases import HFDatasetLoader
```

### **2. Se configura el dataset y la salida**

```python
loader = HFDatasetLoader( dataset_name="BrianSuToronto/california_housing", split="train", filename= "data/california_housing.csv")
```

### **3. Se descarga el dataset y se crea el DataFrame**

```python
df = loader.load()
df.head(3)
```

### **4. Revisamos la estructura del DataFrame**

```python
print(f"Filas: {len(df)}")
print(f"Columnas: {len(df.columns)}")
df.info()
```

### **5. Importamos la clase de reportes**

```python
from clases import Reportes
```

### **6. Apartir de aquí hacemos el renombrado de columnas**

Mostramos los nombres originales de las columnas 

```python
print(df.columns)
```

Les asignamos nuevos nombres

```python
df_transformado = df.rename(columns={
    "MedInc": "Ingreso_mediano",
    "HouseAge": "Edad_vivienda",
    "AveRooms": "Promedio_habitaciones",
    "AveBedrms": "Promedio_dormitorio_hogar",
    "Population": "Poblacion",
    "AveOccup": "Promedio_ocupacion",
    "Latitude": "Latitud",
    "Longitude": "Longitud",
    "MedHouseVal": "Valor_mediano_vivienda"
})
```

**Explicación:** Se renombraron las columnas del DataFrame como lo vemos a continuación con el `DataFrame.rename()`. El resultado se paso a `df_transformado` y se paso a la clase `Reportes` que al final genero el archivo `reporte.txt` con los datos ya transformados.

```python
print(df.columns)               # Nombre original
print(df_transformado.columns)  # Nombre nuevo asignado
```

### **7. Generamos el reporte**

```python
reportes = Reportes()
texto = reportes.generar_reporte(df_transformado)

print(texto)
```

### **8.Imprimimos el reporte**

```python
reportes = Reportes()
reportes.imprimir_reporte(df_transformado)                        # muestra el resultado
```

### **9. Guardamos el reporte en el .txt**

```python
ruta = reportes.guardar_reporte(
    df_transformado,"reportes/california_housing.txt",
) 
print(f"El reporte ha sido guardado en: {ruta}")
```
